```python
import pandas as pd
```

```python
# load dataset

df = pd.read_csv('/content/Final.csv')
```

```python
df.head()
```

```python
import pandas as pd

# Display dataset shapes and columns.

df.shape
df.columns
```

```python
# check for missing values.
df.isnull().sum()
```

```python
# User demographics
print(df['gender'].value_counts())
print(df['age_group'].value_counts())
print(df['country'].value_counts())
```

```python
# screen time analysis

print(df["daily_screen_time_minutes"].describe())
print(df.groupby("age_group")["daily_screen_time_minutes"].mean().sort_values(ascending=False))
```

```python
# platform popularity

print(df["primary_platform"].value_counts())
```

```python
# content preference
print(df["preferred_content_type"].value_counts())
print(df.groupby("primary_device")["preferred_content_type"].value_counts())
```

```python
# Followers distribution
print(df[["user_id","followers_count"]].sort_values("followers_count", ascending=False).head(10))
print(df["followers_count"].describe())
```

```python
# Engagement analysis

print(df.groupby("primary_platform")["engagement_rate_pct"].mean().sort_values(ascending=False))
print(df.groupby("is_content_creator")["engagement_rate_pct"].mean())
```

```python
# social commerce

print(df["has_purchased_via_social"].value_counts(normalize=True) * 100)
print(df.groupby("income_bracket")["monthly_social_spending_usd"].mean().sort_values(ascending=False))
```

```python
# churn risk analysis
df["churn_risk"] = ((df["daily_screen_time_minutes"] < 30) &
                    (df["monthly_social_spending_usd"] == 0))
print("Churn risk users:", df["churn_risk"].sum())
print(df[df["churn_risk"] == True]["age_group"].value_counts())
```

```python
# influence impact
print(df.groupby("follows_influencers")[["monthly_social_spending_usd","engagement_rate_pct"]].mean())
print(df.groupby("follows_influencers")["platform_satisfaction"].value_counts(normalize=True))
```

```python
# cyber bulling
print(df["experienced_cyberbullying"].value_counts())
print(df.groupby("gender")["experienced_cyberbullying"].mean() * 100)
print(df.groupby("age_group")["experienced_cyberbullying"].mean() * 100)
```

```python
# mental health impact
print(df["self_reported_mental_health_effect"].value_counts())
print(df.groupby("addiction_level_1_to_10")["sleep_hours_per_night"].mean())
```

```python
df.to_csv('finaldataset.csv',index=False)
```
